### Q7 - Sum of weights

```python
asia_cars = car_fuel[car_fuel['origin'] == 'Asia']
```

```python
asia_cars.head()
```

```python
asia_cars[['vehicle_weight','model_year']]
```

```python
vehcile_model = asia_cars[['vehicle_weight','model_year']].reset_index(drop=True).head(7)
print(vehcile_model)
```

```python
X = vehcile_model.to_numpy()
X
```

```python
XTX = X.T @ X
print(XTX)
```

```python
X_inv = np.linalg.inv(XTX)
print(X_inv)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
inv_T = X_inv @ X.T
print(inv_T)
```

```python
w =  inv_T @ y 
print(w)
```

```python
w.sum()
```
